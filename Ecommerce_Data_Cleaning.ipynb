# E-commerce Data Cleaning - Simple Python Notebook

This notebook cleans the final enriched e-commerce datasets.

Files used:
- customers.csv
- orders.csv
- order_items.csv
- products.csv
- payments.csv
- returns.csv

The code is kept simple and uses the actual column names in these files.


## Step 1: Import libraries

```python
import pandas as pd
import numpy as np
```

## Step 2: Load all datasets

```python
customers = pd.read_csv("customers.csv")
orders = pd.read_csv("orders.csv")
order_items = pd.read_csv("order_items.csv")
products = pd.read_csv("products.csv")
payments = pd.read_csv("payments.csv")
returns = pd.read_csv("returns.csv")
```

## Step 6: Check data types

```python
print(customers.dtypes)
print(orders.dtypes)
print(order_items.dtypes)
print(products.dtypes)
print(payments.dtypes)
print(returns.dtypes)
```

## Step 7: Check missing values

```python
print("Customers")
print(customers.isnull().sum())

print("\nOrders")
print(orders.isnull().sum())

print("\nOrder Items")
print(order_items.isnull().sum())

print("\nProducts")
print(products.isnull().sum())

print("\nPayments")
print(payments.isnull().sum())

print("\nReturns")
print(returns.isnull().sum())
```

## Step 8: Check duplicate rows

```python
print("Customers duplicates:", customers.duplicated().sum())
print("Orders duplicates:", orders.duplicated().sum())
print("Order Items duplicates:", order_items.duplicated().sum())
print("Products duplicates:", products.duplicated().sum())
print("Payments duplicates:", payments.duplicated().sum())
print("Returns duplicates:", returns.duplicated().sum())
```

## Step 9: Remove exact duplicate rows

```python
customers = customers.drop_duplicates()
orders = orders.drop_duplicates()
order_items = order_items.drop_duplicates()
products = products.drop_duplicates()
payments = payments.drop_duplicates()
returns = returns.drop_duplicates()
```

## Step 10: Clean column names

```python
customers.columns = customers.columns.str.strip().str.lower().str.replace(" ", "_")
orders.columns = orders.columns.str.strip().str.lower().str.replace(" ", "_")
order_items.columns = order_items.columns.str.strip().str.lower().str.replace(" ", "_")
products.columns = products.columns.str.strip().str.lower().str.replace(" ", "_")
payments.columns = payments.columns.str.strip().str.lower().str.replace(" ", "_")
returns.columns = returns.columns.str.strip().str.lower().str.replace(" ", "_")
```

## Step 11: Remove extra spaces from text columns

```python
for col in customers.select_dtypes(include="str"):
    customers[col] = customers[col].str.strip()

for col in orders.select_dtypes(include="str"):
    orders[col] = orders[col].str.strip()

for col in order_items.select_dtypes(include="str"):
    order_items[col] = order_items[col].str.strip()

for col in products.select_dtypes(include="str"):
    products[col] = products[col].str.strip()

for col in payments.select_dtypes(include="str"):
    payments[col] = payments[col].str.strip()

for col in returns.select_dtypes(include="str"):
    returns[col] = returns[col].str.strip()
```

## Step 12: Convert date columns

```python
customers["registration_date"] = pd.to_datetime(
    customers["registration_date"],
    errors="coerce"
)

orders["order_date"] = pd.to_datetime(
    orders["order_date"],
    errors="coerce"
)

orders["delivery_date"] = pd.to_datetime(
    orders["delivery_date"],
    errors="coerce"
)

payments["payment_date"] = pd.to_datetime(
    payments["payment_date"],
    errors="coerce"
)

returns["return_date"] = pd.to_datetime(
    returns["return_date"],
    errors="coerce"
)
```

## Step 13: Convert numeric columns

```python
customers["age"] = pd.to_numeric(customers["age"], errors="coerce")

order_items["quantity"] = pd.to_numeric(
    order_items["quantity"],
    errors="coerce"
)

order_items["unit_price"] = pd.to_numeric(
    order_items["unit_price"],
    errors="coerce"
)

order_items["discount"] = pd.to_numeric(
    order_items["discount"],
    errors="coerce"
)

products["cost_price"] = pd.to_numeric(
    products["cost_price"],
    errors="coerce"
)

products["selling_price"] = pd.to_numeric(
    products["selling_price"],
    errors="coerce"
)

products["stock_quantity"] = pd.to_numeric(
    products["stock_quantity"],
    errors="coerce"
)

products["rating"] = pd.to_numeric(
    products["rating"],
    errors="coerce"
)

payments["payment_amount"] = pd.to_numeric(
    payments["payment_amount"],
    errors="coerce"
)

returns["return_quantity"] = pd.to_numeric(
    returns["return_quantity"],
    errors="coerce"
)

returns["refund_amount"] = pd.to_numeric(
    returns["refund_amount"],
    errors="coerce"
)

orders["shipping_cost"] = pd.to_numeric(
    orders["shipping_cost"],
    errors="coerce"
)
```

## Step 14: Check negative values

```python
print("Negative age:", (customers["age"] < 0).sum())
print("Negative quantity:", (order_items["quantity"] < 0).sum())
print("Negative unit price:", (order_items["unit_price"] < 0).sum())
print("Negative discount:", (order_items["discount"] < 0).sum())
print("Negative cost price:", (products["cost_price"] < 0).sum())
print("Negative selling price:", (products["selling_price"] < 0).sum())
print("Negative stock:", (products["stock_quantity"] < 0).sum())
print("Negative rating:", (products["rating"] < 0).sum())
print("Negative payment amount:", (payments["payment_amount"] < 0).sum())
print("Negative return quantity:", (returns["return_quantity"] < 0).sum())
print("Negative refund amount:", (returns["refund_amount"] < 0).sum())
print("Negative shipping cost:", (orders["shipping_cost"] < 0).sum())
```

## Step 15: Check zero quantity and invalid ranges

```python
print("Order items with quantity 0:", (order_items["quantity"] == 0).sum())
print("Ratings above 5:", (products["rating"] > 5).sum())
print("Discounts above 1:", (order_items["discount"] > 1).sum())
print("Age below 18:", (customers["age"] < 18).sum())
print("Age above 100:", (customers["age"] > 100).sum())
```

## Step 16: Fix zero quantity

```python
order_items.loc[order_items["quantity"] == 0, "quantity"] = 1
```

## Step 17: Check important categorical values

```python
print("Order Status")
print(orders["order_status"].value_counts(dropna=False))

print("\nPayment Status")
print(payments["payment_status"].value_counts(dropna=False))

print("\nPayment Method")
print(orders["payment_method"].value_counts(dropna=False))

print("\nCustomer Segment")
print(customers["customer_segment"].value_counts(dropna=False))

print("\nAcquisition Channel")
print(customers["acquisition_channel"].value_counts(dropna=False))
```

## Step 18: Standardize text values

```python
orders["order_status"] = orders["order_status"].str.lower()
orders["payment_method"] = orders["payment_method"].str.lower()
payments["payment_status"] = payments["payment_status"].str.lower()
customers["gender"] = customers["gender"].str.lower()
customers["customer_segment"] = customers["customer_segment"].str.lower()
customers["acquisition_channel"] = customers["acquisition_channel"].str.lower()
```

## Step 19: Check categories after standardization

```python
print("Order Status")
print(orders["order_status"].value_counts(dropna=False))

print("\nPayment Status")
print(payments["payment_status"].value_counts(dropna=False))

print("\nPayment Method")
print(orders["payment_method"].value_counts(dropna=False))
```

## Step 20: Fill missing values

```python
customers["gender"] = customers["gender"].fillna("unknown")
customers["city"] = customers["city"].fillna("unknown")

order_items["unit_price"] = order_items["unit_price"].fillna(
    order_items["unit_price"].median()
)

order_items["discount"] = order_items["discount"].fillna(0)

orders["shipping_city"] = orders["shipping_city"].fillna("unknown")

returns["refund_amount"] = returns["refund_amount"].fillna(
    returns["refund_amount"].median()
)
```

### Note about delivery_date

`delivery_date` can remain missing for cancelled orders because a cancelled order does not have an actual delivery date. This is a valid missing value, not necessarily a data error.


## Step 21: Check missing values after cleaning

```python
print("Customers missing values:", customers.isnull().sum().sum())
print("Orders missing values:", orders.isnull().sum().sum())
print("Order Items missing values:", order_items.isnull().sum().sum())
print("Products missing values:", products.isnull().sum().sum())
print("Payments missing values:", payments.isnull().sum().sum())
print("Returns missing values:", returns.isnull().sum().sum())
```

## Step 22: Check duplicate IDs

```python
print("Duplicate customer IDs:", customers["customer_id"].duplicated().sum())
print("Duplicate order IDs:", orders["order_id"].duplicated().sum())
print("Duplicate order item IDs:", order_items["order_item_id"].duplicated().sum())
print("Duplicate product IDs:", products["product_id"].duplicated().sum())
print("Duplicate return IDs:", returns["return_id"].duplicated().sum())
```

## Step 23: Remove duplicate customer IDs

```python
customers = customers.drop_duplicates(
    subset=["customer_id"],
    keep="first"
).reset_index(drop=True)
```

## Step 24: Check relationships between tables

```python
print(
    "Orders with invalid customer IDs:",
    (~orders["customer_id"].isin(customers["customer_id"])).sum()
)

print(
    "Order items with invalid order IDs:",
    (~order_items["order_id"].isin(orders["order_id"])).sum()
)

print(
    "Order items with invalid product IDs:",
    (~order_items["product_id"].isin(products["product_id"])).sum()
)

print(
    "Payments with invalid order IDs:",
    (~payments["order_id"].isin(orders["order_id"])).sum()
)

print(
    "Returns with invalid order IDs:",
    (~returns["order_id"].isin(orders["order_id"])).sum()
)

print(
    "Returns with invalid product IDs:",
    (~returns["product_id"].isin(products["product_id"])).sum()
)
```

## Step 25: Check date ranges

```python
print("Customer registration date:")
print(customers["registration_date"].min(), "to", customers["registration_date"].max())

print("\nOrder date:")
print(orders["order_date"].min(), "to", orders["order_date"].max())

print("\nDelivery date:")
print(orders["delivery_date"].min(), "to", orders["delivery_date"].max())

print("\nPayment date:")
print(payments["payment_date"].min(), "to", payments["payment_date"].max())

print("\nReturn date:")
print(returns["return_date"].min(), "to", returns["return_date"].max())
```

## Step 26: Check delivery dates

```python
valid_delivery_dates = (
    orders["delivery_date"].notna()
    & (orders["delivery_date"] < orders["order_date"])
)

print(
    "Delivery dates before order dates:",
    valid_delivery_dates.sum()
)
```

## Step 27: Final dataset sizes

```python
print("Customers:", customers.shape)
print("Orders:", orders.shape)
print("Order Items:", order_items.shape)
print("Products:", products.shape)
print("Payments:", payments.shape)
print("Returns:", returns.shape)
```

## Step 28: Save cleaned datasets

```python
customers.to_csv("customers_clean.csv", index=False)
orders.to_csv("orders_clean.csv", index=False)
order_items.to_csv("order_items_clean.csv", index=False)
products.to_csv("products_clean.csv", index=False)
payments.to_csv("payments_clean.csv", index=False)
returns.to_csv("returns_clean.csv", index=False)

print("All cleaned datasets saved successfully.")
```
